### Flattening the curve

‘Flattening the Curve’
Data visualizations such as thisone are an efficient and effectiveform of communicating publichealth information, but not toeveryone
Activity: In what ways is this datavisualization inaccessible, and towhom? Let’s discuss!

While the "Flattening the Curve" graph is good, it definitely leaves some folks behind. Since it's a static image, anyone relying on a screen reader is completely locked out of the information unless there's a highly detailed text alternative provided. Visually, the sideways text on the y-axis is a headache for people with dyslexia or mobility issues who can't easily tilt their screens, and that faint, gray dotted line showing hospital capacity is really tough to spot if you have low vision.

Beyond the visuals, the graph assumes a lot about its audience's background. Because it's a purely abstract concept with no actual numbers, it can be confusing for folks who need concrete data to make sense of a chart. Plus, it leans heavily on medical jargon like "uncontrolled transmission" and "epidemic peak." Swapping those out for everyday language would go a long way in making this crucial public health info accessible to people with lower health literacy.

This chart tracks the daily routines of eleven famous creatives over a 24-hour period, running from midnight to midnight. A color key breaks down their activities. Teal for sleep, red for creative work, green for day jobs or admin, yellow for food and relaxing, dark blue for exercise, and grey for everything else.

Digging into the details, most of the group got a normal six to eight hours of sleep. Honoré de Balzac is the wild card, splitting his rest and pulling a massive seven-hour writing shift at night. Day jobs also vary wildly. Kant and Vonnegut logged heavy admin hours, while Murakami and Beethoven had none. Interestingly, barely anyone bothered to schedule a workout.

Zooming out, you can spot a huge cluster of red in the mornings. Almost everyone tackled their main creative work between 6 AM and noon. Except for extreme night owls like Balzac and Voltaire, most stuck to a standard day-night schedule. Pacing differs heavily, though: Murakami used massive, unbroken blocks of time, while Vonnegut bounced between tasks all day long.

Ultimately, this shows there's no magic routine for success. Seeing early birds next to night owls proves that putting in the hours matters more than the actual clock. It’s also refreshing to see historical giants logging admin time, reminding us they had real bills to pay. Plus, those giant blocks of afternoon downtime prove that relaxing isn't a waste—it's what fueled their best creative work.



# 08 Data Viz As Advocacy 

### Persuasive visualization: Brooks

Activity
How did this diagram of the Brooks use rational, moral, and emotional appeal to make a caseto its audiences?

Rational
It uses the cold "facts" and precise scale of an architectural blueprint. By laying out exact measurements and cross-sections, it logically proves that "humane" conditions were physically impossible.

Moral
By framing the diagram under the "Regulated Slave Trade Act," it exposes the stark hypocrisy of the law. It appeals to a basic sense of right and wrong by showing humans treated entirely as stacked commercial cargo.

Emotional
The stark, claustrophobic visual of bodies packed shoulder-to-shoulder triggers immediate feelings of suffocation and panic. It forces viewers to viscerally imagine the horror and misery of the conditions without relying on gory details.

### Activity: comparing data visualizations

**How are they different? What does each visualization ‘do’?**
The bar graph shows means and some type of interval around the mean. Likely 95% confidence intervals, but maybe something else (e.g.,1 SD). So, it shows summary statistics. The actual distribution is unsure; virtually many, many combinations of values could make up those summary statistics. 
In comparison, the stripplot shows the actual distribution of the use of gendered words on rate my professor. 


**What are the pros and cons of each?**
The pros of the bar graph is that it shows a high-level depicition of the result. It clearly showcases the difference in means without needing to manually calculate anything. The con is that the raw data is not depicted, so there is uncertainty around the precise distribution.
For the stripplot, the distribution of words by discipline is much clearer. However, a summary of means is more obuscated.

```python
#plotly
import plotly.graph_objects as go
x1 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y1 = np.array([110, 180, 240, 99, 220])
```

```python
# make plot
graph = go.Figure()
graph.add_trace(go.Bar(x=x1, y=y1))

# include title
graph.update_layout(title="Pirate Scores", xaxis_title="Pirates", yaxis_title="Score")
graph.show()

# save as html
graph.write_html("pirategraph.html")
```

```python
# customize plotly graph
graph = go.Figure()
graph.add_trace(go.Scatter(x=x1, y=y1, mode='markers', marker=dict(size=15, color='hotpink', opacity=1, line=dict(width=5, color='purple'))))
graph.update_layout(title='Interactive Pirate Plot', xaxis_title='Pirates', yaxis_title='Scores', width=500, height=500)
```

```python
# wordclouds
from wordcloud import WordCloud
df = pd.read_csv("https://raw.githubusercontent.com/prasertcbs/basic-dataset/master/movie_quotes.csv", on_bad_lines='skip')
print(df)
```

```python

text = " ".join(each for each in df.quote)
wordcloud = WordCloud(background_color="white", colormap='inferno').generate(text)
fig, ax = plt.subplots(figsize=(7,3))
ax.imshow(wordcloud, interpolation='bilinear')
ax.axis("off")
```

```python

from matplotlib_venn import venn2, venn2_circles, venn2_unweighted
A = set(["apple", "banana", "watermelon"])
B = set(["pumpkin", "blueberry", "apple", "key lime"])
```

```python

diagram = venn2_unweighted([A, B], set_labels=('Fruits', 'Pies'), set_colors=("blue", "red"), alpha=0)
plt.show()


diagram.get_label_by_id("10").set_text("\n".join(A - B))
diagram.get_label_by_id("11").set_text("\n".join(A & B))
diagram.get_label_by_id("01").set_text("\n".join(B - A))
plt.show()
```
