### Slide 10: Objective vs. Neutral Data Visualization
Activity: Discuss what information we can learn from visualizations and if they are objective/neutral.

Even when two visualizations show the exact same dataset, the creator's design choices (axis scales, chart types, color highlighting) have potential to change the narrative. Therefore, data visualizations are rarely entirely "objective" or "neutral"—they always reflect a subjective choice about what the creator wants the audience to focus on. Additionally, the specific data selected to vizualize is also a strategic choice by the visualizer.
