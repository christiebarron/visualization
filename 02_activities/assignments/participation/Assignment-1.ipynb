```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100,50)
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y)
```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.bar(x,y)
13
```

```python
# line chart
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
```

```python
# histogram
fig, ax = plt.subplots(figsize=(5, 3))
ax.hist(y)
```

```python
# adding 
ax.set_title('Total growth over time')
ax.set_ylabel('Total growth')
ax.set_xlabel('Years since start')
fig.tight_layout()
fig
```

```python
# modifying labels
font1 = {'family':'sans-serif','color':'blue','size':20}
font2 = {'family':'monospace','color':'green','size':14}

# modifying levels in chart
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig.tight_layout()
fig
```

```python
# moving labels and titles
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
fig
```

```python
# revise the scatterplot markers
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,
           y,
          marker='*',         # NEW ADDITIONS
          color = "indigo")   # NEW ADDITIONS
fig.show()

```

```python
# modifying lines too
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,
        y,
        marker='*', 
        color = 'indigo', 
        linestyle = '--',   # NEW ADDITIONS
        linewidth = 2)      # NEW ADDITIONS
fig.show()
```

```python
# revising colour here
fig, ax = plt.subplots(figsize=(5, 3))
# color being the new addition
ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
fig.show()
```

```python
# add gridlines
fig, ax = plt.subplots(figsize=(5, 3))
# color being the new addition
ax.plot(x,y,marker='*', color = '#7425b9', linestyle = '--', linewidth = 2)
fig.show()
```
